Programa para detectar si una palabra es Palíndromo

```python
palabra = input("Escribe la palabra a analizar")
palabra = palabra.upper()
lista = list(palabra)
print(lista)
lista.reverse()
print(lista)
sep = ""
inversion = sep.join(lista)
print(inversion)
if inversion == palabra:
    print("La palabra es un Palíndromo")
else:
   print("La palabra no es un Palíndromo")
```

```python
persona = dict(Nombre = "Juan", Edad = 65, Ciudad = "CDMX")  #Crear un diccionario con la función dict
```

```python
print(persona)
```

```python
persona["Genero"] = "Masculino"      #Agregar un elemento a un diccionario
print(persona)
```

```python
for key in persona:   #Obtener las keys
    print(key)
```

```python
for valores in persona.values():     #Obtener los valores de los elementos del diccionario
    print(valores)
```

```python
for llave, valor in persona.items():  #Obtener tanto las llaves como los valores
    print (llave, valor)
```
