Métodos (Listas y Cadenas)

Los métodos son "funciones" asociadas a un tipo de objeto en Python.

```python
profe = "Fernando"
profe[3]
```

```python
listapil = ["El Bosque","Las Aguilas","Presidentes","Olivar del Conde","Isidro Fabela"]
print(listapil)
```

```python
listapil.append("Metiche")   # el método append agrega un elemento al final de la lista.
```

```python
print(listapil)
```

```python
# Cuando se necesita agregar un elemento en una posición distinta al final de la lista, se utiliza el método insert

listapil.insert(1,"elemento1")
print(listapil)
```

```python
animales = ["Perro","Gato","Hamster"]
```

```python
listapil.append(animales)
print(listapil)
```

```python
# el método append agregó la lista como un conjunto.
listapil[7]
```

```python
listapil[7][2]
```

```python
# El método extend me permite agregar los elementos de una lista a otra como elementos independientes.
listapil.extend(animales)
print(listapil)
```

```python
listapil.remove("Presidentes")
print(listapil)
```

```python
listapil[6].remove("Perro") # eliminar el elemento perro dentro de la lista animales, que aparece como elemento 6
print(listapil)
```

```python
listapil.remove(animales)   # eliminar la lista animales de la lista listapil
print(listapil)
```

```python
prueba = [1,1,2,2,3,3]
print(prueba)
```

```python
# El método sort es sensible a las mayúsculas y las ordena antes que a las minúsculas.
# reverse, invierte el orden.
# función len (), devuelve el tamaño de la cadena que se coloque entre los paréntesis.
```

```python
listapil.sort(key = len)   # el parámetro key cambia la oredenación de la lista según la función especificada.
print(listapil)
```

```python
listapil.sort(reverse = True) # El método reverse nos devuelve la lista invertida.
print(listapil)
```

```python
print(listapil)
listapil.reverse()
print(listapil)
```

```python
copialistapil = listapil.copy()  # El método copy hace una copia real de la lista.
copy2listapil = listapil         # Al copiar la lista de esta manera, solamente se crea un acceso directo de la lista lista original,
                                # por lo que al modificar la lista utiizando el acceso directo se modifica la lista original.
```

```python
print(copialistapil)
print(copy2listapil)
```

```python
copialistapil.reverse()
print(copialistapil)
print(listapil)
```

```python
copy2listapil.reverse()
print(copy2listapil)
print(listapil)
```

# Métodos para Cadenas

```python
Cadena = "Hoy es sábado"
```

```python
Cadena.lower()
```

```python
Cadena.capitalize()
```

```python
Cadena.upper()
```

```python
print(Cadena)
Cadena.find("sábado")
Cadena.find("es")
```

```python
# Los métodos de cadenas no modifican la lista original.
# Cuando en un bloque de código hay mas de una instrucción que genere salida sin el comando print, únicamente se mostrará la ultima salida.
```

```python
print(Cadena)
print(len(Cadena))
Cadena.startswith("es",4, 13)
```

```python
print(Cadena)
print(len(Cadena))
Cadena.endswith("sábado")
```

```python
print(Cadena)
print(len(Cadena))
Cadena1 = Cadena.replace("sábado","domingo")
print(Cadena1)
```

```python
Palabras = Cadena.split(" ")
print(Palabras)
```

```python
Palabras.append("por la tarde.")
print(Palabras)
```

```python
separador =" "
Frase = separador.join(Palabras)
Frase1 = " ".join(Palabras)
print("\n" + Frase, "\n" + Frase1)  # "\n" significa cambio de línea, el signo + es para indicar que se va a concatenar.
```
