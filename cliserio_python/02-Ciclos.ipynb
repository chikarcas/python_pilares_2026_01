** CICLO IF **

```python
x=3                                  # declarando el valor de la variable x
if x == 3:                            # verifica si el valor de la variable es 3
    print("x es igual a 3")
else:                                 # Que hacer en caso de que el valor de la variable no sea 3
    print("x no es igual a 3")
```

```python
# Este código revisa si un número n está en el intervalo (0,1)  0 < n < 1
n=-1
if n <= 0:
    print("n es menor o igual que 0")
elif n >= 1:
    print("n es mayor o igual que 1")
else:
    print("n está entre 0 y 1")
```

```python
n = float(input("Introduce un número"))
if n < 0:
    print("n es menor que 0")
elif n > 1:
    print("n es mayor que 1")
else:
    print("n está entre 0 y 1")
```

La función input() siempre devuelve un valor str, por lo que hay que definir previamente que tipo de variable es

```python
# Este programa verifica si un número n está entre 100 y 1000, incluyendolos.

n = float(input("Introduce un número"))
if n < 100:
    print("n es menor que 100")
elif n > 1000:
    print("n es mayor que 1000")
else:
    print("n está entre 100 y 1000")
```
