#LIBRETA DE CONTACTOS

```python
# 1. Inicializamos la libreta de contactos como un diccionario vacío
libreta_contactos = {}

print("¡Bienvenido a tu Libreta de Contactos en Python!")
print("================================================")

# 2. Bucle principal: controla si el usuario quiere seguir interactuando con el programa
continuar = True

while continuar:
    # Preguntamos si desea agregar un contacto
    opcion = input("\n¿Quieres agregar un nuevo contacto? sí (1) no (Cualquier otro valor)")
    
    if opcion == "1":
        print("\n--- Creando Nuevo Contacto ---")
        
        # Cada contacto será un diccionario propio para guardar sus datos organizados
        datos_contacto = {}
        
        # Preguntamos por el Nombre (obligatorio para identificarlo en nuestra libreta)
        nombre = input("Ingresa el nombre del contacto: ").strip()
        
        # Preguntamos si desea agregar Teléfono
        agregar_telefono = input(f"¿Quieres agregar teléfono para {nombre}? sí (1) no (Cualquier otro valor) ")
        if agregar_telefono == "1":
            telefono = input("Introduce el número de teléfono: ").strip()
            datos_contacto["teléfono"] = telefono
            
        # Preguntamos si desea agregar Correo
        agregar_correo = input(f"¿Quieres agregar correo para {nombre}? sí (1) no (Cualquier otro valor) ")
        if agregar_correo == "1":
            correo = input("Introduce el correo electrónico: ").strip()
            datos_contacto["correo"] = correo
            
        # 3. Guardamos el contacto en nuestra libreta general usando el nombre como clave (key)
        libreta_contactos[nombre] = datos_contacto
        print(f"¡Contacto '{nombre}' guardado con éxito!")
        
    else:
        # Si dice que no, rompemos el bucle cambiando la variable de control
        continuar = False
        print("\nSaliendo del programa...")
            
# 4. Al finalizar el bucle, mostramos el resultado final (ideal para análisis de datos)
print("\n================================================")
print("Tu Libreta de Contactos Final:")
print(libreta_contactos)
```
